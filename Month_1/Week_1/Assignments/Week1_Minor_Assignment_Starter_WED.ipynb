1. Write a function called `calculate_total_price(price, quantity, tax_rate=0.08)` that:
   - Calculates the subtotal (price × quantity)
   - Adds tax based on the provided rate
   - Returns the final total
   - Test your function with at least 3 different scenarios

````python
# Your code here
def calculate_total_price(price, quantity, tax_rate=0.08):
    """
    Calculate the total price including tax.

    Parameters:
        price (float): Price of a single item
        quantity (int): Number of items
        tax_rate (float): Tax rate (default is 0.08 or 8%)

    Returns:
        float: Final total price including tax
    """
    subtotal = price * quantity
    tax = subtotal * tax_rate
    total = subtotal + tax
    return round(total, 2)

# Test scenarios
print("Test 1:", calculate_total_price(50.0, 2))               # Expected: 108.0
print("Test 2:", calculate_total_price(19.99, 3, 0.1))         # Expected: 65.97
print("Test 3:", calculate_total_price(120.5, 1, 0.07))        # Expected: 128.14
````

2. Write a function called `analyze_sales(products_data)` that:
   - Takes a list of product dictionaries
   - Groups products by category
   - Calculates the average price per category
   - Returns a dictionary with categories as keys and average prices as values
   - Test your function with a sample list of at least 6 products across 3 categories

````python
# Your code here
def analyze_sales(products_data):
    """
    Analyze sales by grouping products by category and calculating average price per category.

    Parameters:
        products_data (list): List of product dictionaries with 'category' and 'price' keys.

    Returns:
        dict: A dictionary where keys are categories and values are average prices.
    """
    from collections import defaultdict

    category_totals = defaultdict(float)
    category_counts = defaultdict(int)

    for product in products_data:
        category = product['category']
        price = product['price']
        category_totals[category] += price
        category_counts[category] += 1

    avg_prices = {
        category: round(category_totals[category] / category_counts[category], 2)
        for category in category_totals
    }

    return avg_prices


# Sample data to test
products_sample = [
    {"name": "Bluetooth Headphones", "price": 59.99, "category": "Electronics"},
    {"name": "Wireless Mouse", "price": 25.49, "category": "Electronics"},
    {"name": "Office Chair", "price": 139.99, "category": "Furniture"},
    {"name": "Standing Desk", "price": 299.99, "category": "Furniture"},
    {"name": "Notebook", "price": 3.99, "category": "Stationery"},
    {"name": "Pen Set", "price": 7.50, "category": "Stationery"},
]

# Run the test
result = analyze_sales(products_sample)
print("Average price per category:")
for category, avg_price in result.items():
    print(f"{category}: ${avg_price}")
````

3. Write a function called `check_inventory(product, threshold=10)` that:
   - Returns "Low stock" if quantity is below the threshold
   - Returns "In stock" if quantity is at or above the threshold
   - Returns "Out of stock" if quantity is 0
   - Test with different product scenarios

````python
# Your code here
def check_inventory(product, threshold=10):
    """
    Checks the inventory status of a product based on quantity and threshold.

    Parameters:
        product (dict): A dictionary containing at least a 'quantity' key.
        threshold (int): The minimum quantity considered "in stock".

    Returns:
        str: Inventory status ("Out of stock", "Low stock", or "In stock")
    """
    quantity = product.get("quantity", 0)

    if quantity == 0:
        return "Out of stock"
    elif quantity < threshold:
        return "Low stock"
    else:
        return "In stock"


# Test cases
product1 = {"name": "Wireless Mouse", "quantity": 0}
product2 = {"name": "Bluetooth Speaker", "quantity": 5}
product3 = {"name": "Laptop Stand", "quantity": 10}
product4 = {"name": "Webcam", "quantity": 15}

print(f"{product1['name']}: {check_inventory(product1)}")
print(f"{product2['name']}: {check_inventory(product2)}")
print(f"{product3['name']}: {check_inventory(product3)}")
print(f"{product4['name']}: {check_inventory(product4)}")
````
